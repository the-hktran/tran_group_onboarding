# Notebook 02: $\mathcal{O}(N^8)$ vs. $\mathcal{O}(N^5)$ 4-Index AO-to-MO ERI Transformations with `np.einsum`

Transforming the 4-index electron repulsion integral tensor $(\mu\nu|\lambda\sigma)$ from the AO basis to the MO basis in $\mathcal{O}(N^5)$ via quarter transformations.

```python
import numpy as np
rng = np.random.default_rng(42)
eri_ao = rng.normal(size=(10, 10, 10, 10))
C = rng.normal(size=(10, 10))
eri_mo = np.einsum('uvls,up,vq,lr,st->pqrt', eri_ao, C, C, C, C, optimize=True)
print('Transformed ERI shape:', eri_mo.shape)
```
