```python
import numpy as np
from scipy.linalg import eigh

charges = np.array([8.0, 1.0, 1.0])
coords_bohr = np.array([
    [0.00000000,  0.00000000,  0.00000000],
    [0.00000000, -1.43052268,  1.10926933],
    [0.00000000,  1.43052268,  1.10926933],
])
diff = coords_bohr[:, None, :] - coords_bohr[None, :, :]
dist_mat = np.linalg.norm(diff, axis=-1)
iu = np.triu_indices(len(charges), k=1)
e_nuc = float(np.sum((charges[:, None] * charges[None, :])[iu] / dist_mat[iu]))
print(f'Vectorized Water E_nuc: {e_nuc:.8f} Eh')
```
