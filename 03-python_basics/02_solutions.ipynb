## Section 0: Foundations — Variables, Lists, NumPy Arrays, and For Loops

Before diving into quantum mechanical operators and Dirac notation, let's establish the fundamental programming building blocks in Python and NumPy.

### 1. Variables and Types
In Python, a **variable** is a named reference to a value stored in memory. You assign values to variables using the `=` assignment operator:

### 2. Python Lists
A **list** is an ordered, mutable collection of items enclosed in square brackets `[...]`. Elements are accessed using **0-based indexing**:
```python
my_list = [10, 20, 30, 40]
print(my_list[0])    # 10 (the first element)
my_list[1] = 99      # Replaces 20 with 99
```

### 3. NumPy Arrays
While Python lists can store arbitrary objects, numerical scientific computing requires **NumPy arrays** (`np.ndarray`). NumPy arrays:
- Store elements of a uniform data type (e.g., float, complex) in contiguous memory blocks.
- Enable vectorized linear algebra operations (vector addition, scaling, matrix multiplications).

To create an array, pass a Python list to `np.array()`:
```python
arr = np.array([1.0, 2.0, 3.0])
```

Like lists, 1D NumPy arrays use 0-based index brackets `arr[index]`:
```python
# Calling/Accessing an element:
```

## Exercise 2: Commutator, Anticommutator, and Matrix Adjoint

Szabo & Ostlund define the **commutator** and **anticommutator** of two operators (matrices) as:
$$
[A, B] = AB - BA, \qquad \{A, B\} = AB + BA
$$

In Exercise 1.2 of the chapter, you are given:
$$
A = \begin{pmatrix} 1 & 1 & 0 \\ 1 & 2 & 2 \\ 0 & 2 & -1 \end{pmatrix}, \qquad B = \begin{pmatrix} 1 & -1 & 1 \\ -1 & 0 & 0 \\ 1 & 0 & 1 \end{pmatrix}

2. Construct matrices `A` and `B` as 2D NumPy float arrays.
3. Compute `comm_AB = commutator(A, B)` and `anticomm_AB = anticommutator(A, B)`.

````python
def commutator(A, B):
    """Compute [A, B] = AB - BA"""
    return A @ B - B @ A

def anticommutator(A, B):
    """Compute {A, B} = AB + BA"""
    return A @ B + B @ A

### START CODE HERE
# Set up matrices A and B
A = np.array([[1, 1, 0], [1, 2, 2], [0, 2, -1]])
B = np.array([[1, -1, 1], [-1, 0, 0], [1, 0, 1]])

comm_AB = commutator(A, B)
anticomm_AB = anticommutator(A, B)
### END CODE HERE

print(comm_AB)
print(anticomm_AB)
````

## Exercise 3: Testing Hermiticity and Unitarity

Chapter 1 highlights two fundamental classes of matrices in quantum mechanics:
1. **Hermitian (Self-adjoint)**: $A = A^\dagger$, where $A^\dagger = (A^*)^T$. Observables correspond to Hermitian operators.
2. **Unitary**: $U^\dagger U = U U^\dagger = \mathbf{1}$. Transformations between orthonormal bases are unitary.

````python
def is_hermitian(M, tol=1e-8):
    """Check if M == M^dagger"""
    return np.allclose(M, M.conj().T, atol=tol)

def is_unitary(M, tol=1e-8):
    """Check if M^dagger @ M == I"""
    ### START CODE HERE
    return np.allclose(M.conj().T @ M, np.eye(M.shape[0]), atol=tol)
    ### END CODE HERE

### START CODE HERE
# Define sigma_x, sigma_y, and sigma_z
sigma_x = np.array([[0, 1], [1, 0]], dtype=complex)
sigma_y = np.array([[0, -1j], [1j, 0]], dtype=complex)
sigma_z = np.array([[1, 0], [0, -1]], dtype=complex)
### END CODE HERE
````

$$
\mathcal{O} |\alpha\rangle = \omega_\alpha |\alpha\rangle

$$

````python
### START CODE HERE
# 1. Define matrix O
O = np.array([[3, 1], [1, 3]])

# 2. Solve eigenvalue problem using scipy.linalg.eigh
eigenvalues, U = scipy.linalg.eigh(O)

# 3. Compute Omega_reconstructed = U^dagger @ O @ U
Omega_reconstructed = U.conj().T @ O @ U

# 4. Compute completeness sum: sum_alpha (|alpha><alpha|)
identity_reconstructed = np.sum([np.outer(U[:, i], U[:, i].conj()) for i in range(U.shape[1])], axis=0)
### END CODE HERE
````

## Exercise 5: Functions of Matrices & Löwdin Orthogonalization ($S^{-1/2}$)

Section 1.1.7 establishes that for any function $f(A)$ of a diagonalizable matrix $A = U \mathbf{\Omega} U^\dagger$:
$$
f(A) = U f(\mathbf{\Omega}) U^\dagger = U \begin{pmatrix} f(\omega_1) & & 0 \\ & \ddots & \\ 0 & & f(\omega_N) \end{pmatrix} U^\dagger
$$

In quantum chemistry, when atomic basis functions are non-orthogonal, they have an overlap matrix $S_{ij} = \langle \phi_i | \phi_j \rangle$. To form an orthonormal set, **Löwdin symmetric orthogonalization** requires constructing the matrix power $S^{-1/2}$:
$$
S^{-1/2} = U s^{-1/2} U^\dagger
$$
where $s = \mathrm{diag}(s_1, \dots, s_N)$ are the eigenvalues of $S$.

### Your Task:
1. Write a function `lowdin_orthogonalizer(S)` that:
   - Diagonalizes $S$ using `scipy.linalg.eigh` to obtain eigenvalues `s_vals` and eigenvectors `U`.
   - Computes $s^{-1/2} = \mathrm{diag}(s_1^{-1/2}, s_2^{-1/2}, \dots)$.
   - Returns $S^{-1/2} = U s^{-1/2} U^\dagger$.
2. Test it on the overlap matrix:
$$
S = \begin{pmatrix} 1.0 & 0.4 \\ 0.4 & 1.0 \end{pmatrix}
$$
3. Verify that $S^{-1/2} S S^{-1/2} = \mathbf{1}$.

````python
def lowdin_orthogonalizer(S):
    """Compute S^(-1/2) using spectral decomposition"""
    ### START CODE HERE
    e, v = np.linalg.eigh(S)
    s = np.diag(1.0 / np.sqrt(e))
    return v @ s @ v.conj().T
    ### END CODE HERE

### START CODE HERE
# Set up matrix S and compute S_inv_sqrt
S = np.array([[1.0, 0.4], [0.4, 1.0]])
S_inv_sqrt = lowdin_orthogonalizer(S)
### END CODE HERE
````

## Exercise 7: The Linear Variation Method (Rayleigh-Ritz)

Section 1.3 develops the **linear variation method**, one of the most vital foundations of quantum chemistry.
Given a trial wavefunction expanded in a set of linearly independent (but generally **non-orthogonal**) basis functions $\{\phi_i\}_{i=1}^N$:
$$
|\widetilde{\Phi}\rangle = \sum_{i=1}^N c_i |\phi_i\rangle
$$
Minimizing the Rayleigh quotient leads to the **generalized eigenvalue problem** (Eq. 1.164):
$$
\mathbf{H} \mathbf{c} = E \mathbf{S} \mathbf{c}
$$
where:
- $H_{ij} = \langle \phi_i | \hat{\mathcal{H}} | \phi_j \rangle$
- $S_{ij} = \langle \phi_i | \phi_j \rangle$

### Physical Problem:
We will estimate the ground- and excited-state energies of a particle in a 1D box of length $L=1$ ($x \in [0, 1]$, in atomic units $\hbar = m = 1$, where $\hat{\mathcal{H}} = -\frac{1}{2}\frac{d^2}{dx^2}$).

We use a **non-orthogonal polynomial basis** satisfying Dirichlet boundary conditions $\phi_i(0) = \phi_i(1) = 0$:
$$
\phi_i(x) = x^i (1 - x), \quad i = 1, 2, \dots, N
$$
Notice:
$$
\frac{d^2 \phi_j}{dx^2} = \frac{d^2}{dx^2}[x^j - x^{j+1}] = j(j-1) x^{j-2} - (j+1)j x^{j-1}
$$

### Your Task:
1. Define functions `phi(i, x)` and `d2_phi(i, x)` for $i \ge 1$.
2. For $N = 4$ basis functions, compute the $4 \times 4$ matrices $\mathbf{S}$ and $\mathbf{H}$ using `scipy.integrate.quad`:
   - $S_{ij} = \int_0^1 \phi_i(x) \phi_j(x) \, dx$
   - $H_{ij} = \int_0^1 \phi_i(x) \left( -\frac{1}{2} \frac{d^2 \phi_j}{dx^2} \right) dx$
3. Solve the generalized eigenvalue problem $\mathbf{H}\mathbf{c} = E\mathbf{S}\mathbf{c}$ using `scipy.linalg.eigh(H, S)`.
4. Compare your variational eigenvalues against the exact analytical values: $E_n = \frac{n^2 \pi^2}{2}$.

````python
def phi(i, x):
    """Basis function phi_i(x) = x^i * (1 - x) for i = 1, 2, ..."""
    ### START CODE HERE
    return x**i * (1 - x)
    ### END CODE HERE

def d2_phi(i, x):
    """Second derivative of phi_i(x)"""
    ### START CODE HERE
    return i * (i - 1) * x**(i - 2) * (1 - x) - 2 * i * x**(i - 1)
    ### END CODE HERE

### START CODE HERE
N = 4
S_var = np.zeros((N, N))
H_var = np.zeros((N, N))

# Assemble S_var and H_var using scipy.integrate.quad
# Note: i and j loop from 1 to N
for i in range(1, N + 1):
    for j in range(1, N + 1):
        S_var[i - 1, j - 1], _ = scipy.integrate.quad(lambda x: phi(i, x) * phi(j, x), 0, 1)
        H_var[i - 1, j - 1], _ = scipy.integrate.quad(lambda x: phi(i, x) * (-d2_phi(j, x)) / 2, 0, 1)

# Solve the generalized eigenvalue problem H c = E S c using scipy.linalg.eigh(a, b)
energies_var, c_vecs = scipy.linalg.eigh(H_var, S_var)
### END CODE HERE
````
